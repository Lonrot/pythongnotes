## 1. What are Classes?

A **class** is like a blueprint or template for creating objects.
Think of it as a cookie cutter - the class is the cutter, and the objects (instances) are the cookies.

Classes allow you to:
- Group related data (attributes)
- Group related functions (methods)
- Create reusable code structures

```python
# A simple class example
class Dog:
    # This is a class, a blueprint for creating dog objects
    pass  # 'pass' means "do nothing for now"

# Creating an object (instance) from the class
my_dog = Dog()
print("Created a dog:", my_dog)
print("Type:", type(my_dog))
```

## 2. Classes with Attributes

Attributes are variables that belong to a class.
They store data about the object.

```python
class Dog:
    # __init__ is a special method called when creating a new object
    # It's called a "constructor"
    def __init__(self, name, age):
        self.name = name  # Attribute: dog's name
        self.age = age    # Attribute: dog's age

# Creating dog objects with different attributes
dog1 = Dog("Angel", 3)
dog2 = Dog("Ivan", 5)

print(f"Dog 1: {dog1.name}, Age: {dog1.age}")
print(f"Dog 2: {dog2.name}, Age: {dog2.age}")
```

```python
#Ivan practice
class Ivan:
    def __init__ (self, name, height, age):
        self.name = name
        self.height = height
        self.age = age
ivan1 = Ivan('Ivan', '177 cm', 34)
print(f'Person- Name:{ivan1.name}, Height:{ivan1.height}, Age:{ivan1.age}')
```

### Understanding 'self'

`self` refers to the specific instance of the class.
- `self.name` means "this object's name"
- It's like saying "my name" vs "your name" - it refers to the specific object

## 3. What are Methods?

**Methods** are functions that belong to a class.
They define what actions an object can perform.

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    # This is a method - a function inside a class
    def bark(self):
        print(f"{self.name} says: Woof! Woof!")
    
    def get_info(self):
        return f"{self.name} is {self.age} years old"

# Create a dog
my_dog = Dog("Charlie", 2)

# Call methods using dot notation
my_dog.bark()  # Method is invoked here
info = my_dog.get_info()  # Method returns a value
print(info)
```

## 4. Methods Run Only When Invoked

**Important:** Methods do NOT run automatically!
They only execute when you explicitly call them.

Exception: `__init__` runs automatically when creating an object.

```python
class Robot:
    def __init__(self, name):
        self.name = name
        print(f"Robot {self.name} created! (This runs automatically)")
    
    def start(self):
        print(f"{self.name} is starting up...")
    
    def work(self):
        print(f"{self.name} is working...")
    
    def shutdown(self):
        print(f"{self.name} is shutting down...")

# Creating a robot - only __init__ runs
robot = Robot("C3PO")

print("\n--- Robot is created but not doing anything yet ---\n")

# Methods only run when we call them
robot.start()     # Now start() is invoked
robot.work()      # Now work() is invoked
robot.shutdown()  # Now shutdown() is invoked
```

## 5. Optional Method Execution

You can choose which methods to run and when.
This makes your code flexible!

```python
class Calculator:
    def __init__(self):
        self.result = 0
        print("Calculator ready!")
    
    def add(self, a, b):
        self.result = a + b
        print(f"Addition: {a} + {b} = {self.result}")
        return self.result
    
    def subtract(self, a, b):
        self.result = a - b
        print(f"Subtraction: {a} - {b} = {self.result}")
        return self.result
    
    def multiply(self, a, b):
        self.result = a * b
        print(f"Multiplication: {a} × {b} = {self.result}")
        return self.result
    
    def divide(self, a, b):
        if b != 0:
            self.result = a / b
            print(f"Division: {a} ÷ {b} = {self.result}")
            return self.result
        else:
            print("Error: Cannot divide by zero!")
            return None

# Create calculator
calc = Calculator()

print("\n--- Choosing which methods to run ---\n")

# We can choose to only use some methods
calc.add(10, 5)
calc.multiply(4, 3)

# We never called subtract or divide - they didn't run!
```

## 6. Conditional Method Execution

You can use if statements to decide whether to run a method.

```python
class Restaurant:
    def __init__(self, customer_name):
        # This ALWAYS runs automatically when creating a Restaurant object
        self.customer_name = customer_name
        self.has_ordered = False
        self.bill = 0
        print(f"Welcome {self.customer_name}!")

    def order_food(self):
        # This ONLY runs when you call it
        print(f"{self.customer_name} ordered food - 15 pounds")
        self.has_ordered = True
        self.bill += 15

    def order_drink(self):
        # This ONLY runs when you call it
        print(f"{self.customer_name} ordered a drink - 3 pounds")
        self.bill += 3

    def order_dessert(self):
        # This ONLY runs when you call it
        print(f"{self.customer_name} ordered dessert - 5 pounds")
        self.bill += 5

    def pay_bill(self):
        # This ONLY runs when you call it
        print(f"Total bill: {self.bill} pounds")


# Example 1: Customer orders everything
print("--- Example 1: Full meal ---")
customer1 = Restaurant("Ivan")
customer1.order_food()
customer1.order_drink()
customer1.order_dessert()
customer1.pay_bill()


# Example 2: Customer skips dessert
print("\n--- Example 2: No dessert ---")
customer2 = Restaurant("Sarah")
customer2.order_food()
customer2.order_drink()
# We don't call order_dessert() so it never runs
customer2.pay_bill()


# Example 3: Using a condition to decide
print("\n--- Example 3: Conditional ordering ---")
customer3 = Restaurant("Alex")

customer3.order_food()

# This variable controls if the method runs
wants_drink = True

if wants_drink:
    customer3.order_drink()  # Only runs if wants_drink is True
else:
    print("No drink ordered")

customer3.pay_bill()


# Example 4: Multiple conditions
print("\n--- Example 4: Multiple conditions ---")
customer4 = Restaurant("Maria")

is_hungry = True
is_thirsty = False
wants_dessert = True

if is_hungry:
    customer4.order_food()

if is_thirsty:
    customer4.order_drink()
else:
    print("No drink needed")

if wants_dessert:
    customer4.order_dessert()

customer4.pay_bill()
```

## 7. How to Consume (Use) Classes

"Consuming" a class means creating instances and using their methods.
Here's the complete workflow:

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        print(f"Account created for {self.owner}")
    
    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited £{amount}. New balance: £{self.balance}")
        else:
            print("Deposit amount must be positive!")
    
    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds!")
        elif amount > 0:
            self.balance -= amount
            print(f"Withdrew £{amount}. New balance: £{self.balance}")
        else:
            print("Withdrawal amount must be positive!")
    
    def get_balance(self):
        print(f"Current balance: £{self.balance}")
        return self.balance

# Step 1: Create an instance (object) of the class
ivan_account = BankAccount("Ivan", 100)

print("\n--- Using the account ---\n")

# Step 2: Call methods on the instance
ivan_account.get_balance()  # Check balance
ivan_account.deposit(50)    # Deposit money
ivan_account.withdraw(30)   # Withdraw money
ivan_account.get_balance()  # Check balance again

# You can create multiple instances
print("\n--- Creating another account ---\n")
jane_account = BankAccount("Jane", 200)
jane_account.get_balance()
```

```python
#Ivan Practice

class Pills:
    def __init__(self, names, total_quantity):
        self.names = names
        self.total_quantity = total_quantity
        self.total_red_pill = 0
        self.total_blue_pill = 0
        print(f'There are {self.total_quantity} pills for {self.names}, red and blue.')
    def red(self, total_red_pill):
        if total_red_pill > self.total_quantity:
            print(f'Not enough available for {self.names}!')
        elif total_red_pill <= self.total_quantity:
            self.total_quantity -= total_red_pill
            print(f'{self.names} has taken {total_red_pill} red pill.')
        else:
            print(f'{self.names} did not take any!')
    def blue(self, total_blue_pill):
        if total_blue_pill > self.total_quantity:
            print(f'Not enough available for {self.names}!')
        elif total_blue_pill <= self.total_quantity:
            self.total_quantity -= total_blue_pill
            print(f'{self.names} has taken {total_blue_pill} blue pill.')
        else:
            print(f'{self.names} did not take any!')
    def Left_over(self):
        print(f'{self.total_quantity} left!')

Ivan = Pills('Ivan', 2)
Ivan.red(1)
Ivan.blue(1)
Ivan.Left_over()


```

## 8. Methods with Parameters

Methods can accept parameters just like regular functions.

```python
class Greeter:
    def __init__(self, name):
        self.name = name
    
    def greet(self, person_name):
        # This method takes a parameter
        print(f"{self.name} says: Hello, {person_name}!")
    
    def greet_multiple(self, *names):
        # This method takes multiple parameters
        for person in names:
            print(f"{self.name} says: Hello, {person}!")

greeter = Greeter("Ivan")

# Invoke methods with different parameters
greeter.greet("Alice")
greeter.greet("Bob")

print()
greeter.greet_multiple("Charlie", "Diana", "Eve")
```
